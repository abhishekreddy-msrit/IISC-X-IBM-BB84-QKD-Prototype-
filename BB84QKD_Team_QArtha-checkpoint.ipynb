1. Introduction
This notebook implements the BB84 Quantum Key Distribution (QKD) protocol using Qiskit. The goal is to demonstrate a functional prototype capable of establishing a shared secret key between two parties (Alice and Bob) with information-theoretic security.

Our implementation includes:

Key Generation: Alice preparing and sending qubits.

Measurement: Bob randomly choosing bases and measuring qubits.

Sifting: Alice and Bob publicly comparing bases to find the "sifted key."

Error Estimation: Calculating the Quantum Bit Error Rate (QBER).

We will demonstrate the protocol in three scenarios:

Ideal Case: A perfect, noise-free simulation.

Eavesdropper Attack: An intercept-resend attack by "Eve."

Noisy Channel: Simulating the protocol on a channel with depolarizing noise.

Finally, we collect and plot the QBER under different noise conditions to analyze the protocol's performance.

2. Setup: Imports
First, we import all necessary libraries from Qiskit, Qiskit Aer (for simulation and noise), and helper libraries for plotting and randomization.

```python
# --- Imports ---

# Qiskit imports
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from qiskit_aer.noise import NoiseModel, depolarizing_error

# Helper imports
import random
import matplotlib.pyplot as plt
import numpy as np

print("Libraries imported successfully!")
```

3. Core Protocol Functions
Here we define the functions for Alice (the sender), Bob (the receiver), and Eve (the eavesdropper).

3.1. Alice: Qubit Preparation
Alice generates her random bits and corresponding random bases (0 for Z-basis, 1 for X-basis). She then encodes these bits into a list of quantum circuits.

```python
def alice_prepares_qubits(num_qubits):
    """
    Alice generates her random bits and bases.
    Returns: (list[int] bits, list[int] bases)
    """
    alice_bits = [random.randint(0, 1) for _ in range(num_qubits)]
    alice_bases = [random.randint(0, 1) for _ in range(num_qubits)] # 0=Z, 1=X
    return alice_bits, alice_bases

def encode_qubits(bits, bases):
    """
    Creates a list of single-qubit circuits based on Alice's bits and bases.
    """
    circuits = []
    for bit, basis in zip(bits, bases):
        qc = QuantumCircuit(1, 1) # 1 qubit, 1 classical bit
        
        # 1. Prepare the bit (apply X if bit is 1)
        if bit == 1:
            qc.x(0)
            
        # 2. Encode in the basis (apply H if basis is 1 (X-basis))
        if basis == 1:
            qc.h(0)
            
        circuits.append(qc)
    return circuits
```

3.2. Bob: Qubit Measurement
Bob generates his own list of random bases and measures the circuits he receives from Alice.

Python

```python
def bob_chooses_bases(num_qubits):
    """
    Bob generates his random bases for measurement.
    """
    bob_bases = [random.randint(0, 1) for _ in range(num_qubits)] # 0=Z, 1=X
    return bob_bases

def measure_qubits(circuits, bases, backend):
    """
    Bob adds his measurement operations to Alice's circuits and runs them.
    """
    measured_bits = []
    
    for qc, basis in zip(circuits, bases):
        # Apply H if Bob wants to measure in X-basis
        if basis == 1:
            qc.h(0)
        
        # Measure in the Z-basis
        qc.measure(0, 0)
        
        # Simulate the circuit
        t_qc = transpile(qc, backend)
        job = backend.run(t_qc, shots=1)
        result = job.result()
        counts = result.get_counts()
        
        # The result is '0' or '1'
        measured_bit = int(list(counts.keys())[0])
        measured_bits.append(measured_bit)
        
    return measured_bits
```

3.3. Public Channel: Sifting
Alice and Bob publicly compare their basis choices and discard any measurements where the bases did not match. This forms the "sifted key."

```python
def sift_keys(alice_bases, bob_bases, alice_bits, bob_results):
    """
    Alice and Bob compare bases and keep only the matching ones.
    """
    sifted_key_alice = []
    sifted_key_bob = []
    
    for i in range(len(alice_bases)):
        if alice_bases[i] == bob_bases[i]:
            # Bases match! Keep these bits.
            sifted_key_alice.append(alice_bits[i])
            sifted_key_bob.append(bob_results[i])
            
    return sifted_key_alice, sifted_key_bob

def calculate_qber(sifted_alice, sifted_bob):
    """
    Calculates the Quantum Bit Error Rate (QBER) from the sifted keys.
    """
    errors = 0
    for a, b in zip(sifted_alice, sifted_bob):
        if a != b:
            errors += 1
            
    qber = errors / len(sifted_alice) if sifted_alice else 0
    return qber
```

3.4. Eve: Intercept-Resend Attack
We define a function for Eve to perform an intercept-resend attack. She will measure Alice's qubits in a random basis and send newly prepared qubits to Bob.

```python
def eve_intercept_resend(circuits, backend):
    """
    Eve intercepts, measures in a random basis, and resends qubits.
    
    Returns:
        (list[QuantumCircuit], list[int]): A list of new circuits for Bob
                                           and the bases Eve used.
    """
    eve_bases = [random.randint(0, 1) for _ in range(len(circuits))]
    new_circuits_for_bob = []

    for qc, eve_basis in zip(circuits, eve_bases):
        # 1. Eve adds her measurement
        if eve_basis == 1: # X-basis
            qc.h(0)
        qc.measure(0, 0)
        
        # 2. Eve simulates the circuit to get her result
        t_qc = transpile(qc, backend)
        job = backend.run(t_qc, shots=1)
        result = job.result()
        eve_bit = int(list(result.get_counts().keys())[0])
        
        # 3. Eve creates a NEW circuit for Bob based on her result
        new_qc = QuantumCircuit(1, 1)
        if eve_bit == 1:
            new_qc.x(0)
        if eve_basis == 1: # Prepare in the same basis she measured
            new_qc.h(0)
            
        new_circuits_for_bob.append(new_qc)
        
    return new_circuits_for_bob, eve_bases
```

4. Simulation 1: Ideal Protocol (No Eve, No Noise)
We first run the protocol on an ideal, noise-free simulator with no eavesdropper. The QBER should be 0.

```python
# --- Simulation 1: Ideal Case ---

NUM_QUBITS = 100
ideal_simulator = AerSimulator()

print("--- Running BB84 Protocol: IDEAL CASE ---")

# 1. Alice
alice_bits, alice_bases = alice_prepares_qubits(NUM_QUBITS)
alice_circuits = encode_qubits(alice_bits, alice_bases)
print(f"Alice's bits:   {alice_bits[:20]}...")
print(f"Alice's bases:  {alice_bases[:20]}... (0=Z, 1=X)")

# 2. Bob
bob_bases = bob_chooses_bases(NUM_QUBITS)
bob_results = measure_qubits(alice_circuits, bob_bases, ideal_simulator)
print(f"Bob's bases:    {bob_bases[:20]}...")
print(f"Bob's results:  {bob_results[:20]}...")

# 3. Sifting
sifted_alice, sifted_bob = sift_keys(alice_bases, bob_bases, alice_bits, bob_results)
print(f"\nSifted Key (Alice): {sifted_alice}")
print(f"Sifted Key (Bob):   {sifted_bob}")

# 4. Error Check
qber = calculate_qber(sifted_alice, sifted_bob)
print(f"\nSifted key length: {len(sifted_alice)}")
print(f"Quantum Bit Error Rate (QBER): {qber * 100:.2f}%")
print("SUCCESS: The keys match perfectly.")
```

5. Simulation 2: Eavesdropper (Eve) Attack
Now, we add Eve to the channel. She intercepts and resends every qubit. This should introduce a detectable error rate (theoretically ~25%).

```python
# --- Simulation 2: Eavesdropper Attack ---

print("\n--- Running BB84 Protocol: WITH EVE ---")

# 1. Alice
alice_bits_eve, alice_bases_eve = alice_prepares_qubits(NUM_QUBITS)
alice_circuits_eve = encode_qubits(alice_bits_eve, alice_bases_eve)

# 2. Eve intercepts
print("... Eve intercepts, measures, and resends qubits ...")
eve_circuits_for_bob, eve_bases = eve_intercept_resend(alice_circuits_eve, ideal_simulator)

# 3. Bob measures what Eve sent him
bob_bases_eve = bob_chooses_bases(NUM_QUBITS)
bob_results_eve = measure_qubits(eve_circuits_for_bob, bob_bases_eve, ideal_simulator)

# 4. Sifting
print("\n--- Public Sifting (with Eve) ---")
sifted_alice_eve, sifted_bob_eve = sift_keys(alice_bases_eve, bob_bases_eve, alice_bits_eve, bob_results_eve)
print(f"Sifted Key (Alice): {sifted_alice_eve}")
print(f"Sifted Key (Bob):   {sifted_bob_eve}")

# 5. Check for errors
qber_eve = calculate_qber(sifted_alice_eve, sifted_bob_eve)
print(f"\nSifted key length: {len(sifted_alice_eve)}")
print(f"Quantum Bit Error Rate (QBER): {qber_eve * 100:.2f}%")

if qber_eve > 0.1: # Set a security threshold
    print("RESULT: High QBER detected! Eavesdropping detected!")
else:
    print("RESULT: Low QBER. No eavesdropper (or Eve got lucky).")
```

6. Simulation 3: Protocol with Channel Noise
Here, we simulate a realistic channel by creating a NoiseModel with depolarizing_error. This error simulates a qubit randomly losing its state. We run this without Eve.

```python
# --- Simulation 3: Noisy Channel ---

# Create a 5% depolarizing error
ERROR_PROB = 0.05
noise_model = NoiseModel()
depol_error = depolarizing_error(ERROR_PROB, 1)
noise_model.add_quantum_error(depol_error, ['x', 'h'], [0])

# Create a new simulator WITH the noise model
noisy_simulator = AerSimulator(noise_model=noise_model)

print(f"\n--- Running BB84 Protocol with {ERROR_PROB*100}% Noise (No Eve) ---")

# 1. Alice
alice_bits_noise, alice_bases_noise = alice_prepares_qubits(NUM_QUBITS)
alice_circuits_noise = encode_qubits(alice_bits_noise, alice_bases_noise)

# 2. Bob (uses the NOISY simulator)
bob_bases_noise = bob_chooses_bases(NUM_QUBITS)
bob_results_noise = measure_qubits(alice_circuits_noise, bob_bases_noise, noisy_simulator)

# 3. Sifting
print("\n--- Public Sifting (with Noise) ---")
sifted_alice_noise, sifted_bob_noise = sift_keys(alice_bases_noise, bob_bases_noise, alice_bits_noise, bob_results_noise)

# 4. Check for errors
qber_noise = calculate_qber(sifted_alice_noise, sifted_bob_noise)
print(f"\nSifted key length: {len(sifted_alice_noise)}")
print(f"Quantum Bit Error Rate (QBER): {qber_noise * 100:.2f}%")
print("RESULT: Non-zero QBER is expected due to channel noise.")
```

7. Metric Analysis: QBER vs. Noise
Finally, as requested by the prompt, we collect and plot key metrics. We will plot the QBER as a function of channel noise, both with and without an eavesdropper.

```python
# --- Plotting Metrics ---

def run_bb84_simulation(num_qubits, noise_prob, with_eve=False):
    """
    A full simulation wrapper to return QBER.
    (This is a silent version for plotting)
    """
    # Setup simulator
    if noise_prob > 0:
        noise_model = NoiseModel()
        depol_error = depolarizing_error(noise_prob, 1)
        noise_model.add_quantum_error(depol_error, ['x', 'h'], [0])
        simulator = AerSimulator(noise_model=noise_model)
    else:
        simulator = AerSimulator()

    # 1. Alice
    alice_bits, alice_bases = alice_prepares_qubits(num_qubits)
    alice_circuits = encode_qubits(alice_bits, alice_bases)
    
    # 2. Channel (Eve or just noise)
    if with_eve:
        circuits_for_bob, _ = eve_intercept_resend(alice_circuits, simulator)
    else:
        circuits_for_bob = alice_circuits

    # 3. Bob
    bob_bases = bob_chooses_bases(num_qubits)
    bob_results = measure_qubits(circuits_for_bob, bob_bases, simulator)
    
    # 4. Sifting
    sifted_alice, sifted_bob = sift_keys(alice_bases, bob_bases, alice_bits, bob_results)

    # 5. Error Check
    return calculate_qber(sifted_alice, sifted_bob)

# --- Run the experiment ---
# Note: This may take a moment to run.
N_QUBITS_LARGE = 200 # Use more qubits for a smoother graph
noise_levels = np.linspace(0, 0.15, 12) # 0% to 15% noise
qber_results_noise = []
qber_results_eve = []

# Suppress print statements during simulation
import sys, os
original_stdout = sys.stdout
sys.stdout = open(os.devnull, 'w')

print("Running simulations for plotting...")
for prob in noise_levels:
    qber_noise = run_bb84_simulation(N_QUBITS_LARGE, noise_prob=prob, with_eve=False)
    qber_results_noise.append(qber_noise)
    
    qber_eve = run_bb84_simulation(N_QUBITS_LARGE, noise_prob=prob, with_eve=True)
    qber_results_eve.append(qber_eve)

sys.stdout = original_stdout # Re-enable print
print("Simulations complete. Plotting...")

# --- Plotting ---
plt.figure(figsize=(10, 6))
plt.plot(noise_levels * 100, [q * 100 for q in qber_results_noise], 'bo-', label='BB84 (Noise Only)')
plt.plot(noise_levels * 100, [q * 100 for q in qber_results_eve], 'ro-', label='BB84 (with Eve + Noise)')
plt.title('QBER vs. Channel Noise and Eavesdropping')
plt.xlabel('Depolarizing Noise Probability (%)')
plt.ylabel('Quantum Bit Error Rate (QBER) (%)')
plt.legend()
plt.grid(True)
plt.show()
```

Plot Analysis
The plot clearly demonstrates the core principles of BB84:

Noise-Only (Blue Line): The QBER scales roughly linearly with the channel noise. This is the baseline error rate Alice and Bob expect.

Eve + Noise (Red Line): The QBER starts at approximately 25% (when noise is 0%) due to Eve's intercept-resend attack. This attack always introduces a high, detectable error rate, regardless of the channel noise.

 Extension 1: Interactive BB84 "Game"
To demonstrate the core concept of eavesdropper detection in a clear and memorable way, we've built a simple interactive "game" using ipywidgets.

This tool allows you to run a small-scale (10 qubit) key exchange with and without Eve. The output log visually demonstrates how the keys match perfectly in the secure case, and how errors are immediately introduced when Eve is present.

```python
# --- Imports for Interactive Game ---
from ipywidgets import Button, Output, VBox, HBox, Layout
from IPython.display import display, clear_output

# --- Create widgets ---
# Output log
log = Output(layout={'border': '1px solid black', 'height': '300px', 'overflow_y': 'scroll'})

# Buttons
btn_send_10 = Button(description='Run: Send 10 Qubits (No Eve)', button_style='success', layout=Layout(width='auto'))
btn_send_10_eve = Button(description='Run: Send 10 Qubits (WITH EVE)', button_style='danger', layout=Layout(width='auto'))
btn_clear = Button(description='Clear Log', button_style='info', layout=Layout(width='auto'))

# --- Helper functions for the game (g_ = game) ---
# We create new versions of the functions to print to our output log

def g_alice_prepares(num_qubits, log_widget):
    alice_bits = [random.randint(0, 1) for _ in range(num_qubits)]
    alice_bases = [random.randint(0, 1) for _ in range(num_qubits)]
    log_widget.append_stdout(f"Alice Bits:   {alice_bits}\n")
    log_widget.append_stdout(f"Alice Bases:  {alice_bases} (0=Z, 1=X)\n")
    return alice_bits, alice_bases, encode_qubits(alice_bits, alice_bases)

def g_bob_measures(circuits, num_qubits, backend, log_widget):
    bob_bases = [random.randint(0, 1) for _ in range(num_qubits)]
    log_widget.append_stdout(f"Bob Bases:    {bob_bases} (0=Z, 1=X)\n")
    bob_results = measure_qubits(circuits, bob_bases, backend)
    log_widget.append_stdout(f"Bob Results:  {bob_results}\n")
    return bob_bases, bob_results

def g_sift_keys(a_bases, b_bases, a_bits, b_bits, log_widget):
    sifted_alice = []
    sifted_bob = []
    for i in range(len(a_bases)):
        if a_bases[i] == b_bases[i]:
            sifted_alice.append(a_bits[i])
            sifted_bob.append(b_bits[i])
    log_widget.append_stdout(f"\n--- Sifting ---\n")
    log_widget.append_stdout(f"Sifted Alice: {sifted_alice}\n")
    log_widget.append_stdout(f"Sifted Bob:   {sifted_bob}\n")
    return sifted_alice, sifted_bob

def g_check_errors(sifted_alice, sifted_bob, log_widget):
    qber = calculate_qber(sifted_alice, sifted_bob)
    log_widget.append_stdout(f"Sifted Length: {len(sifted_alice)}\n")
    log_widget.append_stdout(f"QBER: {qber * 100:.1f}%\n")
    
    if qber > 0.1: # 10% error threshold
        log_widget.append_stdout("\n>>> RESULT: 🔴 EAVESDROPPER DETECTED! <<<\n")
    else:
        log_widget.append_stdout("\n>>> RESULT: 🟢 Keys match! Channel is secure. <<<\n")

# --- Main game logic ---
def run_game(with_eve=False):
    log.append_stdout("=======================================\n")
    if with_eve:
        log.append_stdout("--- Running Protocol WITH EVE ---\n")
    else:
        log.append_stdout("--- Running Protocol (No Eve) ---\n")
    
    N = 10 # Small number for easy visual comparison
    sim = AerSimulator()
    
    # 1. Alice
    a_bits, a_bases, a_circuits = g_alice_prepares(N, log)
    
    # 2. Channel
    if with_eve:
        log.append_stdout("...Eve intercepts and resends qubits...\n")
        # Use the 'eve_intercept_resend' function from the main notebook
        circuits_for_bob, _ = eve_intercept_resend(a_circuits, sim)
    else:
        log.append_stdout("...Qubits sent directly to Bob...\n")
        circuits_for_bob = a_circuits

    # 3. Bob
    b_bases, b_results = g_bob_measures(circuits_for_bob, N, sim, log)
    
    # 4. Sift & Check
    s_alice, s_bob = g_sift_keys(a_bases, b_bases, a_bits, b_results, log)
    g_check_errors(s_alice, s_bob, log)

# --- Button click handlers ---
def on_run_no_eve(b):
    with log:
        run_game(with_eve=False)

def on_run_with_eve(b):
    with log:
        run_game(with_eve=True)

def on_clear(b):
    log.clear_output()

btn_send_10.on_click(on_run_no_eve)
btn_send_10_eve.on_click(on_run_with_eve)
btn_clear.on_click(on_clear)

# --- Display the game UI ---
buttons = HBox([btn_send_10, btn_send_10_eve, btn_clear])
display(VBox([buttons, log]))
```

Cell for CNOT Attack Helper Functions
We need a few new helper functions to build and measure multi-qubit circuits for this simulation.

```python
def encode_qubit_on_qc(qc, bit, basis, qubit_index):
    """Encodes a single bit/basis onto a specific qubit in an existing circuit."""
    if bit == 1:
        qc.x(qubit_index)
    if basis == 1:
        qc.h(qubit_index)

def measure_qubit_on_qc(qc, basis, qubit_index, cbit_index):
    """Measures a specific qubit in a specific basis, storing in a specific classical bit."""
    if basis == 1:
        qc.h(qubit_index)
    qc.measure(qubit_index, cbit_index)

def simulate_and_get_result(qc, backend):
    """Simulates a circuit and returns the single-shot integer result."""
    t_qc = transpile(qc, backend)
    job = backend.run(t_qc, shots=1)
    result = job.result()
    counts = result.get_counts()
    # The result is a string '0' or '1'
    return int(list(counts.keys())[0])
```

Cell for CNOT Attack Simulation
This cell runs the full simulation. It shows Bob's QBER (the errors Alice and Bob see) and compares it to Eve's key accuracy (how much of the key she successfully stole).

```python
def run_cnot_attack_simulation(num_qubits, backend):
    """
    Simulates the full CNOT collector-resend attack.
    q0 = Alice's qubit
    q1 = Eve's ancilla
    """
    print(f"--- Running Advanced CNOT Attack Simulation ({num_qubits} qubits) ---")
    
    # 1. Alice and Bob generate their (secret) bases and bits
    # (Suppressing print statements for a cleaner log)
    alice_bits = [random.randint(0, 1) for _ in range(num_qubits)]
    alice_bases = [random.randint(0, 1) for _ in range(num_qubits)]
    bob_bases = [random.randint(0, 1) for _ in range(num_qubits)]
    
    bob_results = []
    eve_ancilla_circuits = [] # Eve stores her ancilla circuits to measure LATER

    # 2. Alice sends, Eve intercepts, Bob measures
    print("... Eve entangling qubits and forwarding to Bob ...")
    for i in range(num_qubits):
        # Create circuit for Bob's measurement
        qc_bob = QuantumCircuit(2, 1) # q0=Alice, q1=Eve, c0=Bob
        
        # Alice prepares q0
        encode_qubit_on_qc(qc_bob, alice_bits[i], alice_bases[i], 0)
        
        # Eve's CNOT attack: q0 is control, q1 is target
        qc_bob.cx(0, 1)
        
        # Bob measures q0
        measure_qubit_on_qc(qc_bob, bob_bases[i], 0, 0)
        
        # Run and get Bob's result
        bob_results.append(simulate_and_get_result(qc_bob, backend))
        
        # Now, create the circuit for Eve (which she holds onto)
        qc_eve = QuantumCircuit(2, 1) # q0=Alice, q1=Eve, c0=Eve
        encode_qubit_on_qc(qc_eve, alice_bits[i], alice_bases[i], 0)
        qc_eve.cx(0, 1)
        eve_ancilla_circuits.append(qc_eve)

    # 3. Sifting (Public knowledge)
    print("... Alice and Bob sift keys ...")
    sifted_alice = []
    sifted_bob = []
    sifted_indices = [] # Eve listens for this
    
    for i in range(num_qubits):
        if alice_bases[i] == bob_bases[i]:
            sifted_alice.append(alice_bits[i])
            sifted_bob.append(bob_results[i])
            sifted_indices.append(i)
            
    # 4. Eve measures her stored ancillas
    print("... Eve measures her stored ancillas using the public bases ...")
    sifted_eve_results = []
    for i in sifted_indices:
        # Get the circuit Eve stored
        qc_eve = eve_ancilla_circuits[i]
        
        # Get the basis Alice/Bob agreed on
        correct_basis = alice_bases[i] 
        
        # Eve measures her ancilla (q1) in that basis
        measure_qubit_on_qc(qc_eve, correct_basis, 1, 0)
        
        # Run and get Eve's result
        sifted_eve_results.append(simulate_and_get_result(qc_eve, backend))

    # 5. Analysis
    print("\n--- CNOT Attack Results ---")
    
    # Calculate QBER (Alice vs. Bob)
    qber = calculate_qber(sifted_alice, sifted_bob)
    print(f"Sifted Key Length: {len(sifted_alice)}")
    print(f"**Bob's QBER (Errors): {qber * 100:.2f}%**")
    
    # Calculate Eve's key accuracy
    eve_accuracy = 1.0 - calculate_qber(sifted_alice, sifted_eve_results)
    print(f"**Eve's Key Accuracy: {eve_accuracy * 100:.2f}%**")
    
    if qber > 0.05: # 5% threshold
         print("Result: QBER is high enough to detect Eve.")
    else:
         print("Result: QBER is low, Eve may have gone undetected.")

# --- Run the simulation ---
ideal_sim = AerSimulator()
run_cnot_attack_simulation(num_qubits=200, backend=ideal_sim)
```
